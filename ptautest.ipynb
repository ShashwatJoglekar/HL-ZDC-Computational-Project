```python
## Run analysis

ptaudf = lhe_analysis(output_file_path, 'p1', 'p2', 'tau+', 'tau-')

## Print head
print(ptaudf.head())
```

```python
## Proton 1 Plots

## Initialize lists
p1_px_data = []
p1_py_data = []
p1_pz_data = []
p1_E_data = []

## Extract the px, py, pz, E values from the dataframe
for i in range(len(ptaudf)):
    p1_px_data.append(ptaudf["p1Momentum"].iloc[i][0])
    p1_py_data.append(ptaudf["p1Momentum"].iloc[i][1])
    p1_pz_data.append(ptaudf["p1Momentum"].iloc[i][2])
    p1_E_data.append(ptaudf["p1Momentum"].iloc[i][3])

## Make histogram plots for each list
plt.figure(figsize=(12, 8))
plt.hist(p1_px_data, bins=100, alpha=0.5, label='p1 px')
plt.title('p1 px Histogram')
plt.xlabel('px Value')
plt.ylabel('Counts')
plt.show()

plt.figure(figsize=(12, 8))
plt.hist(p1_py_data, bins=100, alpha=0.5, label='p1 px')
plt.title('p1 py Histogram')
plt.xlabel('py Value')
plt.ylabel('Counts')
plt.show()

plt.figure(figsize=(12, 8))
plt.hist(p1_pz_data, bins=100, alpha=0.5, label='p1 px')
plt.title('p1 pz Histogram')
plt.xlabel('pz Value')
plt.ylabel('Counts')
plt.show()

plt.figure(figsize=(12, 8))
plt.hist(p1_E_data, bins=100, alpha=0.5, label='p1 px')
plt.title('p1 E Histogram')
plt.xlabel('E Value')
plt.ylabel('Counts')
plt.show()
```
