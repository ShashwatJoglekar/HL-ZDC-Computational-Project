```python
## Importing libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
## Text File Processing

## Remove the non-important lines of the lhe file
def process_lhe(file_path, output_path):
    ## Remove lines till </MGGenerationInfo>
    with open(file_path, 'r') as file:
        lines = file.readlines()
    index_counter = 0
    ## Get to </MGGenerationInfo> index
    for line in lines:
        if '</MGGenerationInfo>' in line:
            break
        index_counter += 1
    ## Remove lines till </MGGenerationInfo>
    lines = lines[index_counter+1:]
    ## Now remove lines till </init>
    index_counter = 0
    for line in lines:
        if '</init>' in line:
            break
        index_counter += 1
    lines = lines[index_counter+1:]
    ## Reconstruct a text file from the processed lines
    output_file_path = output_path
    with open(output_file_path, 'w') as file:
        file.writelines(lines)
```

```python
## Test with a test case
input_file = 'mytestdir1_testcase.lhe'
output_file_path = 'mytestdir_testcase_processed.lhe'
process_lhe(input_file, output_file_path)
```

```python
## Data parser
def parse_lhe(file_path):
    ## Parse events using </event>
    with open(file_path, 'r') as f:
        data = f.read()
        events = data.split('</event>')
    ## Store events in a list
    event_data = []
    for event in events:
        event_data.append(event.strip())
    return event_data
```

```python
## Test on testdir file
input_file = 'mytestdir1_testcase.lhe'
## Process the file
process_lhe(input_file, output_file_path)
## Parse the processed file
parse_test = parse_lhe(output_file_path)

## Check the number of events parsed
print(f"Number of events parsed: {len(parse_test)}")

## Check the first event text
print(f"First event text:\n{parse_test[0]}")
```

```python
## Code to further process event text data
def process_event(event):
    ## Split event text into lines
    lines = event.split('\n')
    ## Remove everything till <event>
    index_counter = 0
    for line in lines:
        if '<event>' in line:
            break
        index_counter += 1
    lines = lines[index_counter+1:]
    ## Remove everything from <mgrwt>
    index_counter = 0
    for line in lines:
        if '<mgrwt>' in line:
            break
        index_counter += 1
    lines = lines[:index_counter]
    ## Remove empty lines
    lines = [line for line in lines if line.strip() != '']
    ## Remove first line which contains event metadata
    lines = lines[1:]
    ## Turn remaining lines into text file
    event_text = '\n'.join(lines)
    return event_text
```

```python
## Test on first event
event_process_test = process_event(parse_test[0])
print(f"Processed first event text:\n{event_process_test}")
```

```python
## Events data processer
def process_events(event_list):
    ## Process each event in the list
    processed_events = []
    for event in event_list:
        processed_event = process_event(event)
        processed_events.append(processed_event)
    return processed_events
```

```python
## Test on test events
processed_events_test = process_events(parse_test)

## Check the number of processed events
print(f"Number of processed events: {len(processed_events_test)}")

## Check first process events
print(f"First processed event:\n{processed_events_test[0]}")
```

```python
## Extract relevant data from events

## Subfunction to extra relevant data from a processed particle line
def process_particle_data(particle_line):
    ## Split the text by spaces
    particle_data = particle_line.split()
    ## Turn each element into a number
    particle_data = [float(x) for x in particle_data]
    ## Return the processed data
    return particle_data


def event_data_extractor(processed_event):
    ## Split event text into lines
    lines = processed_event.split('\n')
    ## Remove extra spaces from each line
    lines = [line.strip() for line in lines]
    ## Check if there are exactly 4 lines for the particles, if not return four None values
    if len(lines) != 4:
        return None, None, None, None
    input_particle_1_line = lines[0]
    input_particle_2_line = lines[1]
    output_particle_1_line = lines[2]
    output_particle_2_line = lines[3]
    ## Process each particle's data
    input_particle_1_data = process_particle_data(input_particle_1_line)
    input_particle_2_data = process_particle_data(input_particle_2_line)
    output_particle_1_data = process_particle_data(output_particle_1_line)
    output_particle_2_data = process_particle_data(output_particle_2_line)
    ## Return the processed data
    return input_particle_1_data, input_particle_2_data, output_particle_1_data, output_particle_2_data
```

```python
## Test case for first processed event
gamma_1_data, gamma_2_data, tau_data, taubar_data = event_data_extractor(processed_events_test[0])

## Print the extracted data
print(f"Gamma 1 data: {gamma_1_data}")
print(f"Gamma 2 data: {gamma_2_data}")
print(f"Tau data: {tau_data}")
print(f"Tau bar data: {taubar_data}")
```

```python
## Extract momentum four-vector from a particles data

def extract_momentum_fourvector(particle_data):
    ## Get each component
    px = particle_data[6]
    py = particle_data[7]
    pz = particle_data[8]
    energy = particle_data[9]
    ## Return as a numpy array
    return np.array([px, py, pz, energy])
```

```python
## Test on tau particle data
tau_momentum = extract_momentum_fourvector(tau_data)
print(f"Tau momentum four-vector: {tau_momentum}")
```

```python
## Convert all the data into a dataframe
def generate_events_dataframe(events, inparticle1, inparticle2, outparticle1, outparticle2):
    ## Initialize lists
    event_index = []
    input_particle_1_momentum = []
    input_particle_2_momentum = []
    output_particle_1_momentum = []
    output_particle_2_momentum = []
    ## Loop through each event
    for i, event in enumerate(events):
        ## Extract relevant data
        inparticle1_data, inparticle2_data, outparticle1_data, outparticle2_data = event_data_extractor(event)
        ## If the result is None, skip the event
        if inparticle1_data is None or inparticle2_data is None or outparticle1_data is None or outparticle2_data is None:
            print(f"Skipping event {i + 1} due to missing particle data.")
            continue
        else:
                    ## Extract momentum four-vectors
            inparticle1_momentum = extract_momentum_fourvector(inparticle1_data)
            inparticle2_momentum = extract_momentum_fourvector(inparticle2_data)
            outparticle1_momentum = extract_momentum_fourvector(outparticle1_data)
            outparticle2_momentum = extract_momentum_fourvector(outparticle2_data)
            ## Append momentum four-vectors to lists
            input_particle_1_momentum.append(inparticle1_momentum)
            input_particle_2_momentum.append(inparticle2_momentum)
            output_particle_1_momentum.append(outparticle1_momentum)
            output_particle_2_momentum.append(outparticle2_momentum)
            ## Append event index
            event_index.append(i + 1)
    ## Create a dataframe
    df = pd.DataFrame({
        'Event Index': event_index,
        inparticle1 + 'Momentum': input_particle_1_momentum,
        inparticle2 + 'Momentum': input_particle_2_momentum,
        outparticle1 + 'Momentum': output_particle_1_momentum,
        outparticle2 + 'Momentum': output_particle_2_momentum
    })
    return df
```
