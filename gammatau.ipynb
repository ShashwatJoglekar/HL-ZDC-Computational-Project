```python
## Run test on pp-tautau~ data

## Data
input_file_path = 'Events.lhe'
output_file_path = 'Events_processed.lhe'
process_lhe(input_file_path, output_file_path)

## Confirm test case files match
```

```python
## Run analysis

gammataudf = lhe_analysis(output_file_path, 'gamma1', 'gamma2', 'tau+', 'tau-')

## Print head
print(gammataudf.head())
```

```python
## Gamma 1 Plots

## Initialize lists
gamma1_px_data = []
gamma1_py_data = []
gamma1_pz_data = []
gamma1_E_data = []

## Extract the px, py, pz, E values from the dataframe
for i in range(len(gammataudf)):
    gamma1_px_data.append(gammataudf["gamma1Momentum"].iloc[i][0])
    gamma1_py_data.append(gammataudf["gamma1Momentum"].iloc[i][1])
    gamma1_pz_data.append(gammataudf["gamma1Momentum"].iloc[i][2])
    gamma1_E_data.append(gammataudf["gamma1Momentum"].iloc[i][3])

## Make histogram plots for each list
plt.figure(figsize=(12, 8))
plt.hist(gamma1_px_data, bins=200, alpha=0.5, label='gamma1 px')
plt.title('gamma1 px Histogram')
plt.xlabel('px Value')
plt.ylabel('Counts')
plt.show()

plt.figure(figsize=(12, 8))
plt.hist(gamma1_py_data, bins=200, alpha=0.5, label='gamma1 py')
plt.title('gamma1 py Histogram')
plt.xlabel('py Value')
plt.ylabel('Counts')
plt.show()

plt.figure(figsize=(12, 8))
plt.hist(gamma1_pz_data, bins=200, alpha=0.5, label='gamma1 pz')
plt.title('gamma1 pz Histogram')
plt.xlabel('pz Value')
plt.ylabel('Counts')
plt.show()

plt.figure(figsize=(12, 8))
plt.hist(gamma1_E_data, bins=200, alpha=0.5, label='gamma1 E')
plt.title('gamma1 E Histogram')
plt.xlabel('E Value')
plt.ylabel('Counts')
plt.show()
```
